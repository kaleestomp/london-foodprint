# Cache Stats Endpoint Test
Calls the backend endpoint that reports in-memory cache totals, live entries, and expired entries.

```python
import json
import requests

LOCAL_URL = "http://localhost:3000"
RENDER_URL = "https://london-explorer.onrender.com"
TIMEOUT_SECONDS = 30

try:
    requests.get(f"{LOCAL_URL}/health", timeout=2).raise_for_status()
    API_BASE = LOCAL_URL
    print(f"Using local backend: {API_BASE}")
except Exception:
    API_BASE = RENDER_URL
    print(f"Using Render backend: {API_BASE}")
```

```python
resp = requests.get(f"{API_BASE}/api/cache/stats", timeout=TIMEOUT_SECONDS)
resp.raise_for_status()
payload = resp.json()
print(json.dumps(payload, indent=2))
```

```python
summary = payload.get("summary", {})
print("Total :", summary.get("total"))
print("Live  :", summary.get("live"))
print("Expired:", summary.get("expired"))
```
