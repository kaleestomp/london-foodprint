#### Run

```python
import pandas as pd
from pathlib import Path
```

```python
RAWS = [f for f in Path('./places').glob('*.csv') if f.is_file()]
RAW_DFS = []
for f in RAWS:
    try: RAW_DFS.append(pd.read_csv(f))
    except: RAW_DFS.append(pd.DataFrame())
df_raw = pd.concat(RAW_DFS, ignore_index=True)
df_raw.drop_duplicates(subset=['id'], inplace=True)
df_raw = df_raw[[
    "id","displayName",
    "primaryTypeDisplayName","primaryType","types",
    "rating","userRatingCount",
    "location","shortFormattedAddress",
    "priceRange","priceLevel",
    "googleMapsUri",
    "websiteUri","businessStatus",
]]
```

```python
df_validity = pd.DataFrame({
    col: df_raw[col].notna().sum() / len(df_raw) for col in df_raw.columns
}, index=[0])
display(df_validity)
df_parsed = df_raw[df_raw['rating'].notna() & df_raw['userRatingCount'].notna()]
```

```python
from type_parsing import get_summary_type, check_takeaway

df_parsed["summaryType"] = df_parsed.apply(get_summary_type, axis=1)
df_parsed["takeaway"] = df_parsed.apply(check_takeaway, axis=1)
# ── Diagnostics ───────────────────────────────────────────────────────────────
dist = df_parsed["summaryType"].value_counts()
unresolved = (df_parsed["summaryType"] == "Restaurant").sum()
print(f"Unique summaryTypes : {dist.nunique()}")
print(f"Still 'Restaurant'  : {unresolved} / {len(df_parsed)}  ({unresolved/len(df_parsed):.1%})")
# display(dist.to_frame("count"))
# for row in df_parsed[df_parsed["summaryType"]=="Ghost Kitchen"].itertuples():
#     display(row.types)
display(df_parsed[df_parsed["takeaway"]!="Dine-In"])
```
