```python
import sys, json
from pathlib import Path
from dotenv import load_dotenv, find_dotenv
import pandas as pd

# Resolve the repository root so the server package is importable from the notebook.
PROJECT_ROOT = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    if (candidate / "server").is_dir():
        PROJECT_ROOT = candidate
        break
if PROJECT_ROOT is None:
    raise RuntimeError("Could not locate the project root containing the server/ directory.")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Load local secrets from a file that is not committed to git.
load_dotenv(find_dotenv(".env.local", usecwd=True))

from server.scripts.adaptive_hexsearch.request_places.request_places import nearby_search
```

```python
COORDINATE = (51.456644, -0.143151)
# --- Example: restaurants within 500 m of Clapham Common ---
results = await nearby_search(
    latitude=COORDINATE[0],
    longitude=COORDINATE[1],
    radius=500.0,
)
```
