```python
import pandas as pd
PATH = r"C:\Users\kylec\Downloads\london_postcodes-ons-postcodes-directory-feb22.csv"
```

```python
PCD = "pcd2"
FIELD = "msoa01"
df = pd.read_csv(PATH, usecols=[PCD, FIELD])
COLS=["E", "EC", "N", "NW", "SE", "SW", "W", "WC"]
df["region-code"] = df[PCD].str.extract(r"^([A-Z]+)")
df=df[df["region-code"].isin(COLS)].reset_index(drop=True)  # Exclude EC area codes

df["borough-code"] = df[PCD].str.extract(r"([A-Z]+\d+)")
df["areacode"] = df[PCD].str.extract(r"(\d+[A-Z]+)$")
df.to_csv("london_postcodes.csv", index=False)
```
