```python
df = pd.read_csv(PATH, usecols=["pcd", "lat", "long"]).rename(columns={"pcd": "postcode"})
```

```python
COLS=["E", "EC", "N", "NW", "SE", "SW", "W", "WC"]
COLS=["SW"]
df["areacode"] = df["postcode"].str.extract(r"^([A-Z]+)")
df["areacode-subcode"] = df["postcode"].str.extract(r"(\d+)")
df["pcd-subcode"] = df["postcode"].str.extract(r"(\d+[A-Z]+)$")
df=df[df["areacode"].isin(COLS)].reset_index(drop=True)  # Exclude EC area codes
df=df[["postcode"]]
df.to_csv("london_postcodes.csv", index=False)
```
